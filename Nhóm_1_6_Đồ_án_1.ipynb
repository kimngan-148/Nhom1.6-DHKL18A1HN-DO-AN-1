## 1. GIỚI THIỆU

Gender Development Index (GDI) là bộ dữ liệu phản ánh mức độ phát triển con người theo khía cạnh giới tính tại các quốc gia trên thế giới. Bộ dữ liệu được sử dụng để nghiên cứu sự khác biệt về mức độ phát triển giữa nam và nữ, đồng thời xem xét sự thay đổi của chỉ số này theo thời gian.

### Mục tiêu và tầm quan trọng của việc phân tích dữ liệu về bất bình đẳng giới trong phát triển con người

Việc phân tích dữ liệu về bất bình đẳng giới trong phát triển con người đóng
vai trò quan trọng trong việc đánh giá mức độ phát triển và bình đẳng giữa nam và nữ tại các quốc gia và khu vực trên thế giới.

Gender Development Index (GDI) là một trong những chỉ số được sử dụng để phản ánh sự khác biệt về mức độ phát triển của con người giữa hai giới thông qua các khía cạnh quan trọng như sức khỏe, giáo dục và thu nhập.

Mục tiêu của việc phân tích dữ liệu GDI là cung cấp cái nhìn tổng quan và chi tiết về khoảng cách phát triển giữa nam và nữ, từ đó xác định những lĩnh vực còn tồn tại sự chênh lệch đáng kể.

Thông qua việc phân tích dữ liệu theo thời gian, chúng ta có thể theo dõi sự thay đổi của mức độ bình đẳng giới giữa các quốc gia và khu vực.

Việc nghiên cứu các chỉ số như tuổi thọ, số năm đi học, thu nhập và GDI giúp xác định những xu hướng phát triển tích cực hoặc những khoảng cách giới vẫn còn tồn tại.

Bên cạnh đó, việc so sánh dữ liệu giữa nam và nữ còn giúp làm rõ sự khác biệt trong cơ hội tiếp cận giáo dục, điều kiện sức khỏe và khả năng tạo ra thu nhập.

## Chuyển đổi dữ liệu sang dạng Tidy Data & Tạo biến mới

```python
import pandas as pd
import numpy as np

# 1. Chuyển dữ liệu từ dạng Rộng (Wide) sang dạng Dài (Long - Tidy Data)
id_cols = ['ISO3', 'Country', 'Continent', 'Hemisphere', 'Human Development Groups', 'UNDP Developing Regions', 'HDI Rank (2021)']
value_cols = [c for c in df.columns if 'Gender Development Index (' in c]

df_long = pd.melt(df, id_vars=id_cols, value_vars=value_cols, var_name='Year_Raw', value_name='GDI')
df_long['Year'] = df_long['Year_Raw'].str.extract(r'(\d{4})').astype(int)
df_long = df_long.drop(columns=['Year_Raw'])

# 2. Mã hóa giá trị khuyết biến định tính
df_long['UNDP Developing Regions'] = df_long['UNDP Developing Regions'].fillna('Developed/Other')
df_long['Human Development Groups'] = df_long['Human Development Groups'].fillna('Unknown')

# 3. Nội suy tuyến tính chuỗi thời gian GDI theo từng quốc gia
df_long = df_long.sort_values(['ISO3', 'Year']).reset_index(drop=True)
df_long['GDI_Clean'] = df_long.groupby('ISO3')['GDI'].transform(
    lambda g: g.interpolate(method='linear', limit=2, limit_direction='both')
)

# 4. Tạo biến mới: Tốc độ tăng trưởng (% GDI Growth)
df_long['GDI_Growth_%'] = df_long.groupby('ISO3')['GDI_Clean'].pct_change() * 100
df_clean = df_long.dropna(subset=['GDI_Clean']).reset_index(drop=True)

# 5. Lưu file Tidy Data chuẩn
df_clean.to_csv("GDI_Cleaned_Tidy.csv", index=False)
print(f"✅ Đã tạo dữ liệu Tidy Data thành công! Kích thước: {df_clean.shape[0]} dòng, {df_clean.shape[1]} cột.")
display(df_clean.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(3, 2, figsize=(16, 18))

# 3.1. Line Chart: Xu hướng GDI 1990 - 2021
sns.lineplot(data=df_clean, x='Year', y='GDI_Clean', hue='Human Development Groups', estimator='mean', errorbar=None, ax=axes[0, 0], linewidth=2.5)
axes[0, 0].set_title('3.1. Xu hướng GDI Trung bình Toàn cầu (1990–2021)', fontweight='bold', fontsize=12)

# 3.2. Histogram + Density: Phân phối & Outliers năm 2021
sns.histplot(data=df_clean[df_clean['Year'] == 2021], x='GDI_Clean', kde=True, color='teal', ax=axes[0, 1])
axes[0, 1].set_title('3.2. Phân phối Chỉ số GDI năm 2021', fontweight='bold', fontsize=12)

# 3.3. Boxplot: So sánh Phân phối GDI theo Vùng UNDP (Phát hiện Outliers)
sns.boxplot(data=df_clean[df_clean['Year'] == 2021], x='UNDP Developing Regions', y='GDI_Clean', palette='Set2', ax=axes[1, 0])
axes[1, 0].set_title('3.3. So sánh Phân phối GDI 2021 theo Phân vùng UNDP', fontweight='bold', fontsize=12)
axes[1, 0].tick_params(axis='x', rotation=20)

# 3.4. Bar Chart: GDI Trung bình theo Châu lục năm 2021
gdi_cont = df_clean[df_clean['Year'] == 2021].groupby('Continent')['GDI_Clean'].mean().reset_index()
sns.barplot(data=gdi_cont, x='Continent', y='GDI_Clean', palette='Blues_d', ax=axes[1, 1])
axes[1, 1].set_title('3.4. Chỉ số GDI Trung bình năm 2021 theo Châu lục', fontweight='bold', fontsize=12)

# 3.5. Grouped Bar Chart: So sánh GDI Châu lục mốc 1990 vs 2021
sns.barplot(data=df_clean[df_clean['Year'].isin([1990, 2021])], x='Continent', y='GDI_Clean', hue='Year', palette='magma', ax=axes[2, 0])
axes[2, 0].set_title('3.5. So sánh GDI giữa các Châu lục (1990 vs 2021)', fontweight='bold', fontsize=12)

# 3.6. Scatter Plot: Mối liên hệ GDI vs HDI Rank 2021
sns.scatterplot(data=df_clean[df_clean['Year'] == 2021], x='HDI Rank (2021)', y='GDI_Clean', hue='Human Development Groups', ax=axes[2, 1], alpha=0.8)
axes[2, 1].set_title('3.6. Mối liên hệ giữa HDI Rank và GDI (2021)', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```
