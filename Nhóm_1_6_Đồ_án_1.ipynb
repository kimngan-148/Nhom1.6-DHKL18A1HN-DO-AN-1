## 1. GIỚI THIỆU

Gender Development Index (GDI) là bộ dữ liệu phản ánh mức độ phát triển con người theo khía cạnh giới tính tại các quốc gia trên thế giới. Bộ dữ liệu được sử dụng để nghiên cứu sự khác biệt về mức độ phát triển giữa nam và nữ, đồng thời xem xét sự thay đổi của chỉ số này theo thời gian.

### Mục tiêu và tầm quan trọng của việc phân tích dữ liệu về bất bình đẳng giới trong phát triển con người

- Việc phân tích dữ liệu về bất bình đẳng giới trong phát triển con người đóng
vai trò quan trọng trong việc đánh giá mức độ phát triển và bình đẳng giữa nam và nữ tại các quốc gia và khu vực trên thế giới. Gender Development Index (GDI) là một trong những chỉ số được sử dụng để phản ánh sự khác biệt về mức độ phát triển của con người giữa hai giới thông qua các khía cạnh quan trọng như sức khỏe, giáo dục và thu nhập. Mục tiêu của việc phân tích dữ liệu GDI là cung cấp cái nhìn tổng quan và chi tiết về khoảng cách phát triển giữa nam và nữ, từ đó xác định những lĩnh vực còn tồn tại sự chênh lệch đáng kể.

- Thông qua việc phân tích dữ liệu theo thời gian, chúng ta có thể theo dõi sự thay đổi của mức độ bình đẳng giới giữa các quốc gia và khu vực. Việc nghiên cứu các chỉ số như tuổi thọ, số năm đi học, thu nhập và GDI giúp xác định những xu hướng phát triển tích cực hoặc những khoảng cách giới vẫn còn tồn tại.

- Bên cạnh đó, việc so sánh dữ liệu giữa nam và nữ còn giúp làm rõ sự khác biệt trong cơ hội tiếp cận giáo dục, điều kiện sức khỏe và khả năng tạo ra thu nhập. Những kết quả này có thể cung cấp cơ sở dữ liệu hữu ích cho việc nghiên cứu các vấn đề xã hội và phát triển con người.

## 5.3. Trực quan hóa nhiều phân phối đồng thời

  + Các nhóm phát triển con người
  + Các khu vực đang phát triển theo UNDP
- Sử dụng Box Plot, Histogram, Violin Plot
## 5.4. Trực quan hoá tỷ lệ

  + Nhóm phát triển con người và khu vực UNDP

  + Khu vực UNDP và mức GDI được phân nhóm

### 2.7. Tiền xử lý dữ liệu

#### 2.7.1. Kiểm tra và xử lý tên cột, dữ liệu dạng chữ

```python
# Chuẩn hóa tên cột
df.columns = df.columns.str.strip()

# Xác định các cột dạng chữ
text_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

# Loại bỏ khoảng trắng thừa ở đầu và cuối giá trị
for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Kiểm tra kết quả
display(df.head())

print("Tên cột sau khi chuẩn hóa:")
print(df.columns.tolist())
```

#### 2.7.2. Chuyển đổi từ Wide sang Long

```python
# Lấy danh sách các cột GDI theo năm
gdi_cols = [col for col in df.columns if col.startswith('Gender Development Index')]

# Trích xuất năm từ tên cột
year_mapping = {col: int(col.replace('Gender Development Index (', '').replace(')', '')) for col in gdi_cols}

# Chuyển đổi Wide sang Long bằng pd.melt
df_long = df.melt(
    id_vars=['ISO3', 'Country', 'Continent', 'Hemisphere', 'Human Development Groups', 'UNDP Developing Regions', 'HDI Rank (2021)'],
    value_vars=gdi_cols,
    var_name='GDI_Year_Col',
    value_name='GDI'
)

# Tạo cột Year kiểu số nguyên
df_long['Year'] = df_long['GDI_Year_Col'].map(year_mapping)
df_long.drop(columns=['GDI_Year_Col'], inplace=True)

# Sắp xếp dữ liệu theo Country và Year
df_long = df_long.sort_values(by=['Country', 'Year']).reset_index(drop=True)

print("Kích thước bảng Wide ban đầu:", df.shape)
print("Kích thước bảng Long sau chuyển đổi:", df_long.shape)
display(df_long.head(10))
```

#### 2.7.3. Lưu dữ liệu sau tiền xử lí

```python
import os
import pandas as pd
import kagglehub

# Đảm bảo dữ liệu đã được tải và gán vào biến df nếu chưa tồn tại trong session
try:
    df
except NameError:
    print("Biến 'df' chưa tồn tại. Đang tải lại dữ liệu từ Kaggle...")
    path = kagglehub.dataset_download("iamsouravbanerjee/gender-development-index-dataset")
    df = pd.read_csv(os.path.join(path, "Gender Development Index.csv"))
    df.columns = df.columns.str.strip()
    text_columns = df.select_dtypes(include=["object"]).columns.tolist()
    for col in text_columns:
        df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

# Lưu file
df.to_csv(
    "GDI_sau_tien_xu_ly.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Đã lưu file GDI_sau_tien_xu_ly.csv thành công!")
```

#**CHƯƠNG 3:** THỐNG KÊ MÔ TẢ

### 3.1. Đại lượng thống kê cơ bản của GDI (2021) và HDI Rank (2021)

```python
# Bảng thống kê đại lượng cơ bản
cols_to_stat = ['Gender Development Index (2021)', 'HDI Rank (2021)']
stats_df = df[cols_to_stat].describe().T

# Tính toán bổ sung Phương sai (Variance) và Trung vị (Median)
stats_df['variance'] = df[cols_to_stat].var()
stats_df['median'] = df[cols_to_stat].median()

# Sắp xếp lại thứ tự cột
stats_df = stats_df[['count', 'mean', 'median', 'std', 'variance', 'min', '25%', '50%', '75%', 'max']]
print("CÁC ĐẠI LƯỢNG THỐNG KÊ CƠ BẢN (NĂM 2021)")
display(stats_df.round(4))
```

### 3.2. Ước lượng tỷ lệ phân bố theo nhóm phát triển con người

```python
# Tần suất và Tỷ lệ phần trăm của Human Development Groups
hd_counts = df['Human Development Groups'].value_counts(dropna=False)
hd_props = (df['Human Development Groups'].value_counts(normalize=True, dropna=False) * 100).round(2)

prop_summary = pd.DataFrame({
    'Số lượng quốc gia': hd_counts,
    'Tỷ lệ (%)': hd_props
})

print("BẢNG PHÂN BỐ CÁC NHÓM PHÁT TRIỂN CON NGUỜI (HUMAN DEVELOPMENT GROUPS)")
display(prop_summary)
```

### 3.3. Ước lượng khoảng và Kiểm định trung bình GDI giữa các nhóm

```python
import scipy.stats as stats

# Chuẩn bị dữ liệu sạch cho ANOVA
df_anova = df.dropna(subset=['Gender Development Index (2021)', 'Human Development Groups'])

# Gom nhóm GDI 2021 theo từng Human Development Group
group_data = [group['Gender Development Index (2021)'].values for name, group in df_anova.groupby('Human Development Groups')]

# Thực hiện kiểm định ANOVA một nhân tố (One-way ANOVA)
f_stat, p_val = stats.f_oneway(*group_data)

print("KIỂM ĐỊNH SO SÁNH GDI TRUNG BÌNH GIỮA CÁC NHÓM PHÁT TRIỂN CON NGƯỜI (2021)")
print(f"Giá trị F-statistic: {f_stat:.4f}")
print(f"Giá trị p-value: {p_val:.4e}")

if p_val < 0.05:
    print("Kết luận: Có sự khác biệt có ý nghĩa thống kê về chỉ số GDI 2021 giữa các nhóm phát triển con người (p < 0.05).")
else:
    print("Kết luận: Không đủ bằng chứng cho thấy sự khác biệt có ý nghĩa thống kê.")
```

#**CHƯƠNG 4:** MỐI LIÊN HỆ GIỮA CÁC BIẾN

### 4.1. Mối liên hệ giữa HDI Rank (2021) và GDI (2021)

```python
# Tính hệ số tương quan Pearson giữa HDI Rank và GDI 2021
corr_hdi_gdi = df['HDI Rank (2021)'].corr(df['Gender Development Index (2021)'])

print(f"Hệ số tương quan Pearson giữa HDI Rank (2021) và GDI (2021): {corr_hdi_gdi:.4f}")
print("Nhận xét: Mối quan hệ nghịch chiều rõ rệt. Thứ hạng HDI càng cao (con số thứ hạng lớn), giá trị GDI càng có xu hướng giảm.")
```

### 4.2. Mối liên hệ tương quan của GDI qua các năm

```python
# Ma trận tương quan giữa các cột GDI từ năm 1990 đến 2021
corr_matrix = df[gdi_cols].corr()

# Lấy hệ số tương quan của tất cả các năm so với năm 2021
corr_with_2021 = corr_matrix['Gender Development Index (2021)'].sort_values(ascending=False)

print("TƯƠNG QUAN CHỈ SỐ GDI CÁC NĂM SO VỚI GDI NĂM 202")
print(corr_with_2021.head(10).round(4))
print("\nCác năm xa nhất (1990 - 1994):")
print(corr_with_2021.tail(5).round(4))
```

### 4.3. Mối liên hệ giữa GDI và Human Development Groups

```python
# Bảng chỉ số GDI 2021 theo từng nhóm Human Development Groups
gdi_by_hd_group = df.groupby('Human Development Groups')['Gender Development Index (2021)'].agg(
    Số_quốc_gia='count',
    GDI_Trung_bình='mean',
    GDI_Trung_vị='median',
    Độ_lệch_chuẩn='std',
    Min='min',
    Max='max'
).reset_index()

print("MỐI LIÊN HỆ GIỮA GDI (2021) VÀ CÁC NHÓM PHÁT TRIỂN CON NGƯỜI")
display(gdi_by_hd_group.round(4))
```
